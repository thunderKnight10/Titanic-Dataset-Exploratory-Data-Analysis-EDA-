### Observation: Missing Value Analysis

The missing-value analysis identifies three columns containing incomplete records in the Titanic dataset.


Age: Contains 177 missing values, representing approximately 19.87% of the dataset.

Cabin: Contains 687 missing values, representing approximately 77.10% of the dataset.

Embarked: Contains 2 missing values, representing approximately 0.22% of the dataset.

The remaining nine columns contain no missing values.

```python
missing_values=df.isna().sum()/len(df)*100
print(missing_values.sort_values(ascending=False))
```

### Observation: Missing Value Percentage Analysis

The missing-value percentage analysis provides a clearer understanding of the proportion of incomplete data in each column.


Cabin: Has the highest percentage of missing values at approximately 77.10%.

Age: Contains approximately 19.87% missing values.

Embarked: Contains approximately 0.22% missing values.

Remaining Columns: All other columns have 0% missing values, indicating complete records.
