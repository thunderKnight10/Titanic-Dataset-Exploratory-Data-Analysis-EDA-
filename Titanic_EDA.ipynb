### Observation: Survival Distribution

The Survived column represents whether a passenger survived the Titanic disaster.

The value counts reveal the following distribution:

Did Not Survive (0): 549 passengers, representing approximately 61.62% of the dataset.

Survived (1): 342 passengers, representing approximately 38.38% of the dataset.

```python
sns.countplot(data=df,x="Survived")
plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of passenger")
plt.show()


```

### Observation: Passenger Survival Distribution

The count plot illustrates the distribution of passengers based on their survival status.

Not Survived (0): 549 passengers.
Survived (1): 342 passengers.

The visualization clearly shows that the number of passengers who did not survive was considerably higher than the number who survived.

```python
pd.crosstab(df["Sex"],
            df["Survived"].replace({0:"Not Survived",1:"Survived"}),
            normalize="index")*100
```

### Observation: Gender-Based Survival Analysis

The cross-tabulation analysis examines the relationship between passenger gender and survival status using percentage distributions.


Female Passengers: Approximately 74.20% survived, while 25.80% did not survive.
Male Passengers: Approximately 18.89% survived, while 81.11% did not survive.

```python
sns.countplot(data=df,x="Sex",hue="Survived")

plt.title("Survival Distribution by Gender")
plt.xlabel("Sex")
plt.ylabel("Number of passenger")
plt.show()
```

### Observation: Survival Distribution by Gender

The count plot illustrates the relationship between passenger gender and survival status by displaying the number of survivors and non-survivors in each gender category.


Female Passengers: The number of female survivors (233) was significantly higher than the number of female non-survivors (81).
Male Passengers: The number of male non-survivors (468) was considerably higher than the number of male survivors (109).

```python
sns.catplot(data=df,
                x="Pclass" , 
                hue="Survived",
                col="Sex",
            kind="count"  )


plt.show()
```

### Observation: Survival Distribution by Passenger Class and Gender

The categorical plot illustrates the relationship between passenger class, gender, and survival status. Separate plots are generated for male and female passengers to make comparisons easier.


First Class: Female passengers had a higher survival count than male passengers. Survival outcomes were comparatively better in this class.

Second Class: Female passengers also showed a higher survival count, while most male passengers did not survive.

Third Class: The number of non-survivors was considerably higher, particularly among male passengers. Female passengers also experienced lower survival counts compared to the higher classes.

```python
corelation=df.corr(numeric_only=True)
corelation
```

### Observation: Correlation Analysis

The correlation matrix examines the linear relationships between numerical features in the Titanic dataset. It helps identify positive and negative associations between variables.


Survived and Pclass: A negative correlation indicates that passengers in higher passenger classes (lower Pclass values) tended to have better survival outcomes.

Survived and Fare: A positive correlation suggests that passengers who paid higher fares tended to have higher survival rates.

SibSp and Parch: A positive correlation indicates that the number of siblings/spouses aboard and parents/children aboard are related.

Age and Pclass: A relatively weak negative relationship suggests some variation in passenger ages across different classes.

```python
sns.heatmap(
            corelation,
            annot=True,
            cmap="coolwarm",
            fmt=".2f"
)

plt.title("Corelation matrix")
plt.show()
```

### Observation: Correlation Heatmap

The correlation heatmap provides a visual representation of the relationships between numerical features in the Titanic dataset. Different colors indicate the direction and strength of correlations, while the numerical annotations display the exact correlation coefficients.


Survived and Pclass: A negative correlation indicates that passengers in first class generally had better survival outcomes than those in lower classes.

Survived and Fare: A positive correlation suggests that passengers who paid higher fares tended to have better survival outcomes.

SibSp and Parch: A positive correlation indicates a relationship between the number of siblings/spouses and parents/children aboard.

Age and Pclass: A negative correlation suggests that passenger age tends to vary inversely with passenger class.

# Final Conclusion

The Exploratory Data Analysis (EDA) of the Titanic dataset provided valuable insights into passenger demographics, survival patterns, data quality, and relationships between different features.

The dataset consists of 891 passenger records and 12 columns. Initial inspection revealed missing values in three columns: `Age`, `Cabin`, and `Embarked`. Among these, `Cabin` had the highest proportion of missing values (77.10%), followed by `Age` (19.87%) and `Embarked` (0.22%). No duplicate records were identified.

Descriptive statistical analysis showed that the average passenger age was approximately 29.70 years, while the average ticket fare was approximately 32.20. The survival distribution revealed that 342 passengers (38.38%) survived, whereas 549 passengers (61.62%) did not survive.

Further analysis of categorical variables revealed a significant difference in survival outcomes between male and female passengers. Approximately 74.20% of female passengers survived, compared to only 18.89% of male passengers. Passenger class analysis also suggested that passengers in higher classes generally had better survival outcomes than those in third class.

The correlation analysis and heatmap provided additional insights into relationships between numerical features. Passenger class showed a negative association with survival, while fare showed a positive association with survival. These findings suggest that passenger class, fare, and gender were important factors associated with survival outcomes.

Overall, this EDA helped identify important patterns, detect data-quality issues, and understand relationships among the dataset's features. These findings provide a strong foundation for the next stages of data preprocessing, feature engineering, and machine learning model development to predict passenger survival.
